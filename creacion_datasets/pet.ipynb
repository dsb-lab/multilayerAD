```python
def visit(visit_str):
    if visit_str == 'bl': #baseline
        return 0
    elif visit_str == 'sc': #screen (normalmente 1 mes antes del baseline)
        return 0
    elif visit_str == 'f': #fail
        return -1
    else:
        try:
            visit_num = visit_str.replace('m', '')
            visit_num = float(visit_num)
            return visit_num
        except:
            return None
```

```python
temp = pd.read_csv('../archivos_originales/pet/UCBERKELEYFDG.csv')
temp
# medidas específicas tomadas de imágenes PET FDG en cinco regiones del cerebro y crea una variable de resumen a partir
# del promedio de estas medidas en ciertas regiones.
temp1 = temp[temp.ROINAME == 'metaroi']
temp2 = temp[temp.ROINAME == 'pons-vermis']

df = temp1[['RID', 'VISCODE2', 'MEAN']].copy()
df.rename(columns={"MEAN": "MEAN_METAROI"}, inplace=True)
df['VISCODE2'] = df['VISCODE2'].astype('str')
df['Visit'] = df['VISCODE2'].apply(visit)
df.drop(columns=['VISCODE2'], inplace=True)

df2 = temp2[['RID', 'VISCODE2', 'MEAN']].copy()
df2.rename(columns={"MEAN": "MEAN_PONSVERMIS"}, inplace=True)
df2['VISCODE2'] = df2['VISCODE2'].astype('str')
df2['Visit'] = df2['VISCODE2'].apply(visit)
df2.drop(columns=['VISCODE2'], inplace=True)

df = pd.merge(df, df2, on=['RID', 'Visit'], how='outer')

df['MEAN_METAROI_PONSVERMIS'] = df['MEAN_METAROI']/df['MEAN_PONSVERMIS']
df.drop(columns=['MEAN_METAROI', 'MEAN_PONSVERMIS'], inplace=True)
```
