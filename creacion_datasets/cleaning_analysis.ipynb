# Análisis crossectional

Tomamos los valores de las primeras visitas. Si alguna de las variables está vacía, tomamos el primer valor no nulo de las posteriores visitas

```python
clinic = pd.read_csv('clinic_v2.csv')
# clinic = clinic[~clinic.ADSP_DX.isna()].copy().reset_index(drop=True)
# RIDs_dx = clinic.RID.tolist()
clinic = clinic.groupby('RID', as_index=False).first()
clinic.drop(columns=['Visit'], inplace=True)
```

```python
gene = pd.read_csv('genetic_v2.csv')
# gene = gene[gene.RID.isin(RIDs_dx)].copy().reset_index(drop=True)
molecular = pd.read_csv('molecular_v2.csv')
# molecular = molecular[molecular.RID.isin(RIDs_dx)].copy().reset_index(drop=True)
molecular = molecular.groupby('RID', as_index=False).first()
molecular.drop(columns=['Visit'], inplace=True)
pet = pd.read_csv('pet_v2.csv')
# pet = pet[pet.RID.isin(RIDs_dx)].copy().reset_index(drop=True)
pet = pet.groupby('RID', as_index=False).first()
pet.drop(columns=['Visit'], inplace=True)
mri = pd.read_csv('mri_v2.csv')
# mri = mri[mri.RID.isin(RIDs_dx)].copy().reset_index(drop=True)
mri = mri.groupby('RID', as_index=False).first()
mri.drop(columns=['Visit'], inplace=True)
risk = pd.read_csv('risk_v2.csv')
# risk = risk[risk.RID.isin(RIDs_dx)].copy().reset_index(drop=True)
risk = risk.groupby('RID', as_index=False).first()
risk.drop(columns=['Visit'], inplace=True)
```

```python
df = gene.join(molecular.set_index('RID'), on='RID', how='outer')
print(len(df.columns))
df = df.join(pet.set_index('RID'), on='RID', how='outer')
print(len(df.columns))
df = df.join(mri.set_index('RID'), on='RID', how='outer')
print(len(df.columns))
df = df.join(risk.set_index('RID'), on='RID', how='outer')
print(len(df.columns))
df = df.join(clinic.set_index('RID'), on='RID', how='outer')
print(len(df.columns))
df.sort_values(by='RID', inplace=True)
df.reset_index(drop=True, inplace=True)
df
```

```python
RIDs = df.RID.unique().tolist()
columnas = df.columns.tolist()
del columnas[0]
print(len(RIDs), len(columnas))
```

```python
cm = np.ones((len(columnas), df.RID.nunique()))
for i in range(len(columnas)):
    for j in range(len(RIDs)):
        if pd.isna(df.loc[j][i+1]):
            cm[i][j] = 0
np.save('cm_todos.npy', cm)
```

```python
df_new = df.copy()
df_new
```

```python
rows_nans = []
for row in range(len(df_new)):
    if df_new.loc[[row]].isna().sum().sum() > 200:
        rows_nans.append(row)
print(f'Hay {len(rows_nans)} filas con más de 200 datos vacíos')
df_new.drop(rows_nans, axis=0, inplace=True)
df_new.reset_index(drop=True, inplace=True)
```

```python
RIDs_def = df_new.RID.unique().tolist()
columnas = df_new.columns.tolist()
del columnas[0]
print(len(RIDs), len(columnas))
cm = np.ones((len(columnas), df_new.RID.nunique()))
for i in range(len(columnas)):
    for j in range(len(RIDs_def)):
        if pd.isna(df_new.loc[j][i+1]):
            cm[i][j] = 0
np.save('cm_todos_limpio.npy', cm)
```
