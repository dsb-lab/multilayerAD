```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
gene = pd.read_csv('genetic.csv')
molecular = pd.read_csv('molecular.csv')
pet = pd.read_csv('pet.csv')
mri = pd.read_csv('mri.csv')
risk = pd.read_csv('risk.csv')
clinic = pd.read_csv('clinic.csv')
```

```python
def filter_df(df, is_gene=False):
    # Empty columns
    empty_columns = []
    for i in range(len(df.columns)):
        cont = 0
        for j in range(len(df)):
            if pd.isna(df[df.columns[i]][j]):
                cont += 1
        if cont == len(df):
            print(f'Numero de nans en {df.columns[i]}: {cont}')
            empty_columns.append(df.columns[i])
    df.drop(empty_columns, axis=1, inplace=True)
    df.drop_duplicates(inplace=True, ignore_index=True)
    df.reset_index(drop=True, inplace=True)
    
    # Empty rows
    rows_nans = []
    for row in range(len(df)):
        if df.loc[[row]].isna().sum().sum() == len(df.columns)-1: #porque RID no tiene ningun nan
            rows_nans.append(row)
    print(f'Hay {len(rows_nans)} filas vacías, las eliminamos')
    df.drop(rows_nans, axis=0, inplace=True)
    df.reset_index(drop=True, inplace=True)    

    if not is_gene:
        df = df[~df.Visit.isna()]
        df = df[df.Visit != -1]
        df.reset_index(drop=True, inplace=True)
        rows_nans = []
        for row in range(len(df)):
            if df.loc[[row]].isna().sum().sum() == len(df.columns)-2: #porque RID y Visit no tiene ningun nan
                rows_nans.append(row)
        print(f'Hay {len(rows_nans)} filas vacías, las eliminamos')
        df.drop(rows_nans, axis=0, inplace=True)
        df.reset_index(drop=True, inplace=True)

    return df

gene = filter_df(gene, is_gene=True)
molecular = filter_df(molecular)
pet = filter_df(pet)
mri = filter_df(mri)
risk = filter_df(risk)
clinic = filter_df(clinic)
```
