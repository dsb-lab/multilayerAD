```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
pd.read_csv('../capas_def/BIOMARKERS.csv')
```

```python
def visit(visit_str):
    if visit_str == 'bl': #baseline
        return 0
    elif visit_str == 'sc': #screen (normalmente 1 mes antes del baseline)
        return 0
    elif visit_str == 'f': #fail
        return -1
    else:
        try:
            visit_num = visit_str.replace('m', '')
            visit_num = float(visit_num)
            return visit_num
        except:
            return None
```

```python
temp = pd.read_csv('../archivos_originales/molecular/ADNI_EUROIMMUN.csv')
df = temp[['RID', 'VISCODE2', 'BETA_AMYLOID_1_42', 'BETA_AMYLOID_42_40']].copy()
df.rename(columns={"BETA_AMYLOID_1_42": "EUR_AB42", 'BETA_AMYLOID_42_40': 'EUR_AB42/40'}, inplace=True)
df['VISCODE2'] = df['VISCODE2'].astype('str')
df['Visit'] = df['VISCODE2'].apply(visit)
df.drop(columns=['VISCODE2'], inplace=True)
```

```python
temp = pd.read_csv('../archivos_originales/molecular/FUJIREBIOABETA.csv')
df2 = temp[['RID', 'VISCODE2', 'ABETA42', 'ABETA42_40']].copy()
df2.rename(columns={"ABETA42": "FUJI_AB42", 'ABETA42_40': 'FUJI_AB42/40'}, inplace=True)
df2['VISCODE2'] = df2['VISCODE2'].astype('str')
df2['Visit'] = df2['VISCODE2'].apply(visit)
df2.drop(columns=['VISCODE2'], inplace=True)

df = pd.merge(df, df2, on=['RID', 'Visit'], how='outer')
```
