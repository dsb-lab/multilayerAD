```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
pd.read_csv('../capas_def/PHENOTYPE.csv')
```

```python
def visit(visit_str):
    if visit_str == 'bl': #baseline
        return 0
    elif visit_str == 'sc': #screen (normalmente 1 mes antes del baseline)
        return 0
    elif visit_str == 'f': #fail
        return -1
    else:
        try:
            visit_num = visit_str.replace('m', '')
            visit_num = float(visit_num)
            return visit_num
        except:
            return None
```

```python
temp = pd.read_csv('../archivos_originales/clinic/MMSE.csv')
df = temp[['RID', 'VISCODE2', 'MMSCORE']].copy()
df.rename(columns={"MMSCORE": "MMSE"}, inplace=True)
df = df[df['MMSE'].notna()]
df['VISCODE2'] = df['VISCODE2'].astype('str')
df['Visit'] = df['VISCODE2'].apply(visit)
df.drop(columns=['VISCODE2'], inplace=True)
```

```python
temp = pd.read_csv('../archivos_originales/clinic/MOCA.csv')
df2 = temp[['RID', 'VISCODE2', 'MOCA']].copy()
df2 = df2[df2['MOCA'].notna()]
df2['VISCODE2'] = df2['VISCODE2'].astype('str')
df2['Visit'] = df2['VISCODE2'].apply(visit)
df2.drop(columns=['VISCODE2'], inplace=True)

df = pd.merge(df, df2, on=['RID', 'Visit'], how='outer')
```
