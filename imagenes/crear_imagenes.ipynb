```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.Graph()

G.add_edge("a", "b", weight=0.6)
G.add_edge("a", "c", weight=0.2)
G.add_edge("c", "d", weight=0.1)
G.add_edge("c", "e", weight=0.7)
G.add_edge("c", "f", weight=0.9)
G.add_edge("a", "d", weight=0.3)

nx.draw_networkx(G, node_size=1500, node_color='skyblue', with_labels=False)
plt.box(False)
plt.tight_layout()
# plt.savefig('netnormal.png', format='png', dpi=600)
plt.show()

elarge = [(u, v) for (u, v, d) in G.edges(data=True) if d["weight"] > 0.5]
esmall = [(u, v) for (u, v, d) in G.edges(data=True) if d["weight"] <= 0.5]

nodes_in_path = nx.shortest_path(G, 'a', 'f', weight='weight')
print(nodes_in_path)
edges_path = [('b', 'a'), ('a', 'c'), ('c', 'e')]

pos = nx.spring_layout(G, seed=7)  # positions for all nodes - seed for reproducibility

# nodes
nx.draw_networkx_nodes(G, pos, node_size=1000, node_color='seagreen')

# edges
nx.draw_networkx_edges(G, pos, width=2, alpha=0.5)
nx.draw_networkx_edges(G, pos, edgelist=edges_path, width=4, alpha=1, edge_color="r")

# node labels
# nx.draw_networkx_labels(G, pos, font_size=20, font_family="sans-serif")
# edge weight labels
# edge_labels = nx.get_edge_attributes(G, "weight")
# nx.draw_networkx_edge_labels(G, pos, edge_labels)

ax = plt.gca()
ax.margins(0.08)
plt.axis("off")
plt.tight_layout()
plt.savefig('path_score.png', format='png', dpi=600)
plt.show()
```

```python
G = nx.Graph()

G.add_edge("A", "B", weight=-0.3)
G.add_edge("A", "C", weight=0.5)

node_inactive = ['A']
edges_positive = [(u, v) for (u, v, d) in G.edges(data=True) if d["weight"] > 0]
edges_negative = [(u, v) for (u, v, d) in G.edges(data=True) if d["weight"] < 0]
edge_labels = dict([((n1, n2), d['weight']) for n1, n2, d in G.edges(data=True)])

pos = nx.circular_layout(G)

# nodes
nx.draw_networkx_nodes(G, pos, node_size=8000, node_color='lightgreen')
nx.draw_networkx_nodes(G, pos, nodelist=node_inactive, node_size=8000, node_color='gray')
nx.draw_networkx_labels(G, pos, font_size=30, font_color='black')

# edges
nx.draw_networkx_edges(G, pos, edgelist=edges_negative, width=2, edge_color="r")
nx.draw_networkx_edges(G, pos, edgelist=edges_positive, width=2, edge_color="b")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color='black', font_size=25)

# Adjust the axis limits
ax = plt.gca()
ax.margins(0.2)
ax.autoscale()

plt.axis("off")
plt.box(False)
plt.tight_layout()
plt.savefig('ex_bool1.png', format='png', dpi=600)
plt.show()


# nodes
nx.draw_networkx_nodes(G, pos, node_size=8000, node_color='lightgreen')
nx.draw_networkx_labels(G, pos, font_size=30, font_color='black')

# edges
nx.draw_networkx_edges(G, pos, edgelist=edges_negative, width=2, edge_color="r")
nx.draw_networkx_edges(G, pos, edgelist=edges_positive, width=2, edge_color="b")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color='black', font_size=25)

# Adjust the axis limits
ax = plt.gca()
ax.margins(0.2)
ax.autoscale()

plt.axis("off")
plt.box(False)
plt.tight_layout()
plt.savefig('ex_bool2.png', format='png', dpi=600)
plt.show()


# nodes
node_inactive = ['B']
nx.draw_networkx_nodes(G, pos, node_size=8000, node_color='lightgreen')
nx.draw_networkx_nodes(G, pos, nodelist=node_inactive, node_size=8000, node_color='gray')
nx.draw_networkx_labels(G, pos, font_size=30, font_color='black')

# edges
nx.draw_networkx_edges(G, pos, edgelist=edges_negative, width=2, edge_color="r")
nx.draw_networkx_edges(G, pos, edgelist=edges_positive, width=2, edge_color="b")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color='black', font_size=25)

# Adjust the axis limits
ax = plt.gca()
ax.margins(0.2)
ax.autoscale()

plt.axis("off")
plt.box(False)
plt.tight_layout()
plt.savefig('ex_bool3.png', format='png', dpi=600)
plt.show()
```
