```python
import pandas as pd
```

```python
df_phen = pd.read_csv('PHENOTYPE.csv')
```

```python
RID_nl = df_phen[df_phen.ADSP_DX == 1].RID.unique()
print(len(RID_nl))
RID_mci = df_phen[df_phen.ADSP_DX == 2].RID.unique()
print(len(RID_mci))
RID_dem = df_phen[df_phen.ADSP_DX == 3].RID.unique()
print(len(RID_dem))
```

```python
for name in ['GENETIC', 'MOLECULAR', 'PET', 'MRI', 'RISKFACTORS', 'PHENOTYPE']:
    print(name)
    df = pd.read_csv(f'{name}.csv')
    df_nl = df[df.RID.isin(RID_nl)]
    df_nl.reset_index(drop=True, inplace=True)
    print('NL', df_nl.shape)
    df_nl.to_csv(f'{name}_nl.csv', index=False)
    df_mci = df[df.RID.isin(RID_mci)]
    df_mci.reset_index(drop=True, inplace=True)
    print('MCI', df_mci.shape)
    df_mci.to_csv(f'{name}_mci.csv', index=False)
    df_dem = df[df.RID.isin(RID_dem)]
    df_dem.reset_index(drop=True, inplace=True)
    print('Dementia', df_dem.shape)
    df_dem.to_csv(f'{name}_dementia.csv', index=False)
```
