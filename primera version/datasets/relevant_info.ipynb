```python
import pandas as pd
```

```python
df = pd.read_csv('RISKFACTORS.csv')
RIDs = df.RID.unique().tolist()
print(len(RIDs))
df_phen = pd.read_csv('PHENOTYPE.csv')
RID_nl = df_phen[df_phen.ADSP_DX == 1].RID.unique()
print(len(RID_nl))
RID_mci = df_phen[df_phen.ADSP_DX == 2].RID.unique()
print(len(RID_mci))
RID_dem = df_phen[df_phen.ADSP_DX == 3].RID.unique()
print(len(RID_dem))
```

```python
def visit(visit_str):
    if visit_str == 'bl': #baseline
        return 0
    elif visit_str == 'sc': #screen (normalmente 1 mes antes del baseline)
        return 0
    elif visit_str == 'f': #fail
        return -1
    else:
        try:
            visit_num = visit_str.replace('m', '')
            visit_num = float(visit_num)
            return visit_num
        except:
            return None
```

```python
df = pd.read_csv('../archivos_originales/risk_factors/PTDEMOG.csv')
df['VISCODE2'] = df['VISCODE2'].astype('str')
df['Visit'] = df['VISCODE2'].apply(visit)
df.drop(columns=['VISCODE2'], inplace=True)
df = df[df.Visit < 1].copy()
df['Date'] = pd.to_datetime(df['USERDATE'])
df['Age'] = df['Date'].dt.year - df['PTDOBYY']
df_nl = df[df.RID.isin(RID_nl)].copy()
print(df_nl.Age.mean())
print(df_nl.Age.std())
df_mci = df[df.RID.isin(RID_mci)].copy()
print(df_mci.Age.mean())
print(df_mci.Age.std())
df_dem = df[df.RID.isin(RID_dem)].copy()
print(df_dem.Age.mean())
print(df_dem.Age.std())
```

```python
df = pd.read_csv('../archivos_originales/clinic/ADSP.csv')
df['VISCODE2'] = df['VISCODE2'].astype('str')
df['Visit'] = df['VISCODE2'].apply(visit)
df.drop(columns=['VISCODE2'], inplace=True)

# Obtener la última visita para cada paciente
ultima_visita = df.groupby('RID')['Visit'].max().reset_index()
# Combinar el resultado con el dataframe original
df_ultima_edad = pd.merge(df, ultima_visita, on=['RID', 'Visit'])
df_nl = df_ultima_edad[df_ultima_edad.RID.isin(RID_nl)].copy()
print(df_nl.AGE.mean())
print(df_nl.AGE.std())
df_mci = df_ultima_edad[df_ultima_edad.RID.isin(RID_mci)].copy()
print(df_mci.AGE.mean())
print(df_mci.AGE.std())
df_dem = df_ultima_edad[df_ultima_edad.RID.isin(RID_dem)].copy()
print(df_dem.AGE.mean())
print(df_dem.AGE.std())

# df[(df.RID.isin(RID_dem)) & (temp.DX != 3)]
print(df[df.RID.isin(RID_dem)].AGE.mean())
print(df[df.RID.isin(RID_dem)].AGE.std())
df
```
