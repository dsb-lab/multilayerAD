```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
temp = pd.read_csv('../archivos_originales/genetic/APOERES.csv')
df = temp[['RID', 'APGEN1', 'APGEN2']].copy()
df.rename(columns={"APGEN1": "APOE_A1", "APGEN2": "APOE_A2"}, inplace=True)
```

```python
def apoe(df):
    if df['APOE_A1'] == 4 and df['APOE_A2'] == 4:
        return 2
    elif df['APOE_A1'] != 4 and df['APOE_A2'] != 4:
        return 0
    else:
        return 1
```

```python
df['APOE'] = df.apply(apoe, axis=1)
```

```python
temp = pd.read_csv('../archivos_originales/genetic/TOMM40.csv')
temp.drop(columns=['PTID', 'update_stamp'], inplace=True)
df = pd.merge(df, temp, on='RID', how='outer')
```

```python
temp = pd.read_csv('../archivos_originales/genetic/DESIKANLAB.csv')
temp.drop(columns=['USERDATE', 'update_stamp'], inplace=True)
df = pd.merge(df, temp, on='RID', how='outer')
df = df.sort_values('RID')
df
```

```python
df.to_csv('genetic.csv', index=False)
```
